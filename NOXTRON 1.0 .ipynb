```python
                                    #NOXTRON 1.0
import webbrowser
import subprocess
from datetime import datetime

# RESPOND FUNCTION

def respond(command):
    command = command.strip().lower()

    # Greetings
    if command in ["hello", "hi", "hey"]:
        return "Greetings. I am NOXTRON. State your command."

    # Identity
    elif command in ["who are you", "what is your name"]:
        return "I am NOXTRON, an advanced personal assistant."

    # Time and date
    elif command == "time":
        return "Current time: " + datetime.now().strftime("%I:%M %p")
    elif command == "date":
        return "Current date: " + datetime.now().strftime("%d %B %Y")

    # Help
    elif command == "help":
        return (
            "Commands:\n"
            "- hello / hi / hey\n"
            "- who are you\n"
            "- time\n"
            "- date\n"
            "- open google\n"
            "- open youtube\n"
            "- open github\n"
            "- open wikipedia\n"
            "- open notepad\n"
            "- open calculator\n"
            "- exit"
        )

    # Open websites
    elif command == "open google":
        webbrowser.open("https://www.google.com")
        return "Accessing Google..."
    elif command == "open youtube":
        webbrowser.open("https://www.youtube.com")
        return "Accessing YouTube..."
    elif command == "open github":
        webbrowser.open("https://www.github.com")
        return "Accessing GitHub..."
    elif command == "open wikipedia":
        webbrowser.open("https://www.wikipedia.org")
        return "Accessing Wikipedia..."

    # Open apps (Windows examples)
    elif command == "open notepad":
        try:
            subprocess.Popen(["notepad.exe"])
            return "Launching Notepad..."
        except Exception:
            return "Unable to launch Notepad on this system."

    elif command == "open calculator":
        try:
            subprocess.Popen(["calc.exe"])
            return "Launching Calculator..."
        except Exception:
            return "Unable to launch Calculator on this system."

    # Exit
    elif command == "exit":
        return "Session terminated."

    else:
        return "Command not recognized. Type 'help' for available commands."

        

# MAIN 
print("NOXTRON initialized. Type 'help' for commands.")

while True:
    user_command = input("You: ")
    answer = respond(user_command)
    print("NOXTRON:", answer)

    if user_command.strip().lower() == "exit":
        break
```
